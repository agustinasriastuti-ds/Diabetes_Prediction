```python
df = pd.read_csv('diabetes.csv')
df.head()
```

```python
print(df.head())
```

```python
#proporsi pasien diabetes dan non-diabetes.
#Keterangan: 0 = Tidak diabetes 1 = Diabetes
plt.figure()
df['Outcome'].value_counts().plot(kind='bar')
plt.xlabel("Outcome")
plt.ylabel("Jumlah Pasien")
plt.title("Distribusi Pasien Diabetes")
plt.show()
```

```python
#Distribusi Usia Pasien
plt.figure()
plt.hist(df['Age'], bins=10)
plt.xlabel("Usia")
plt.ylabel("Jumlah Pasien")
plt.title("Distribusi Usia Pasien")
plt.show()
```

```python
#Distribusi Kadar Glukosa
plt.figure()
plt.hist(df['Glucose'], bins=10)
plt.xlabel("Glucose")
plt.ylabel("Jumlah Pasien")
plt.title("Distribusi Kadar Glukosa")
plt.show()
```

```python
#Hubungan Glukosa terhadap Outcome
#Melihat apakah pasien diabetes memiliki kadar glukosa lebih tinggi.
plt.figure()
plt.scatter(df['Glucose'], df['Outcome'])
plt.xlabel("Glucose")
plt.ylabel("Outcome")
plt.title("Hubungan Glukosa dengan Outcome Diabetes")
plt.show()
```

```python
#Melihat pengaruh berat badan terhadap diabetes.
plt.figure()
plt.scatter(df['BMI'], df['Outcome'])
plt.xlabel("BMI")
plt.ylabel("Outcome")
plt.title("Hubungan BMI dengan Outcome Diabetes")
plt.show()
```

```python
#Korelasi Antar Variabel
#Mengetahui fitur yang paling berpengaruh terhadap diabetes.
corr = df.corr()

plt.figure()
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Heatmap Korelasi Dataset Diabetes")
plt.show()
```
